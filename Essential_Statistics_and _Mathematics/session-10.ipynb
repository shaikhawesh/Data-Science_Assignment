```python
# 2

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


# Prior probability of a restaurant being premium
prior_premium = 0.10

# Probability of discount if restaurant is premium
likelihood_discount_premium = 0.80

# Probability of restaurant being regular
prior_regular = 0.90

# Probability of discount if restaurant is regular
likelihood_discount_regular = 0.20

# Probability of seeing a discount
evidence_discount = (
    likelihood_discount_premium * prior_premium
    + likelihood_discount_regular * prior_regular
)

# Calculate P(Premium | Discount)
posterior = bayes_posterior(
    prior_premium,
    likelihood_discount_premium,
    evidence_discount
)

print("Probability that the restaurant is premium:", posterior)
print("Percentage:", posterior * 100)
```

```python
# 3

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


# Probability that an email is spam
prior_spam = 0.40

# Probability that "free" appears in a spam email
likelihood_free_spam = 0.70

# Probability that an email is not spam
prior_not_spam = 0.60

# Probability that "free" appears in a non-spam email
likelihood_free_not_spam = 0.10

# Probability that an email contains "free"
evidence_free = (
    likelihood_free_spam * prior_spam
    + likelihood_free_not_spam * prior_not_spam
)

# Calculate P(Spam | Free)
posterior = bayes_posterior(
    prior_spam,
    likelihood_free_spam,
    evidence_free
)

print("Probability that the email is spam:", posterior)
print("Percentage:", posterior * 100)
```

```python
# 4

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


# Probability that a user usually likes recommended posts
prior_likes = 0.30

# Probability of liking the post if the user usually likes recommendations
likelihood_likes_post = 0.80

# Probability that the user does not usually like recommendations
prior_not_likes = 0.70

# Probability of liking the post if the user does not usually like recommendations
likelihood_likes_post_not = 0.20

# Probability that the user likes the post
evidence_likes = (
    prior_likes * likelihood_likes_post
    + prior_not_likes * likelihood_likes_post_not
)

# Calculate P(UsuallyLikes | LikesPost)
posterior = bayes_posterior(
    prior_likes,
    likelihood_likes_post,
    evidence_likes
)

print("Probability that the user usually likes recommended posts:", posterior)
print("Percentage:", posterior * 100)
```
