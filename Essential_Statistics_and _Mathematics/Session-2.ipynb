```python
"""
3.
Write a function most_common_rating(ratings) that takes a list of Flipkart
 product ratings (e.g., [5, 4, 4, 3, 5, 4, 2, 4]) and returns the mode 
 (most frequent rating).
"""
def most_common_rating(ratings):
    return st.mode(ratings)

ratings = [5,4,4,3,5,4,2,4]

result = most_common_rating(ratings)

print(result)
```

```python
"""
4.
Given three lists representing YouTube video views for three different
 channels, compare the mean, median, and mode for each channel, and decide
   which channel's data is most affected by outliers.<br><br><em><strong>
   Hint:</strong> Try using one list with a very high value 
   (e.g., [100, 120, 110, 105, 5000]) to see the effect on mean vs median.
   </em>
"""

channel_1 = [100, 120, 110, 105, 5000]
channel_2 = [100, 120, 110, 105, 115]
channel_3 = [100, 100, 110, 120, 100]


def analyze_channel(channel):
    mean = st.mean(channel)
    median = st.median(channel)
    mode = st.mode(channel)

    return mean, median, mode


channels = {
    "Channel 1": channel_1,
    "Channel 2": channel_2,
    "Channel 3": channel_3,
}

for name, data in channels.items():
    mean, median, mode = analyze_channel(data)

    print("\n", name)
    print("Data:", data)
    print("Mean:", mean)
    print("Median:", median)
    print("Mode:", mode)
```
