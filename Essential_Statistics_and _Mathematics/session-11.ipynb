```python
# 3

orders = [
    [10, 20, 15, 12],
    [12, 18, 20, 10],
    [15, 22, 18, 14]
]

dish_totals = [0, 0, 0, 0]

for row in orders:
    for i in range(4):
        dish_totals[i] += row[i]

print("Total orders for each dish:")

for i in range(4):
    print("Dish", i + 1, ":", dish_totals[i])
```

```python
# 4

products = [
    [500, 10],
    [1200, 5],
    [800, 8],
    [1500, 4],
    [300, 15]
]

def calculate_revenue(products):
    revenues = []

    for product in products:
        price = product[0]
        quantity = product[1]

        revenue = price * quantity
        revenues.append(revenue)

    return revenues


revenues = calculate_revenue(products)

print("Revenue for each product:")

for i in range(len(revenues)):
    print("Product", i + 1, ":", revenues[i])
```
