```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=100)


mean = 50
std = 10

x = np.linspace(mean - 4 * std, mean + 4 * std, 200)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.plot(x, y, linewidth=2)

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal Distribution (Mean = 50, SD = 10)")

plt.show()
```

```python
# 2

import numpy as np
from scipy import stats

steps = [
    5000, 5200, 4800, 5100, 5300,
    4900, 5000, 5150, 5050, 4950,
    5100, 5200, 5000, 5050, 4950,
    5150, 5100, 5000, 5050, 5100
]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True).mode[0]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)

if abs(mean - median) < 200 and abs(mean - mode) < 200:
    print("Mean, median, and mode are approximately equal.")
    print("The data may approximately follow a normal distribution.")
else:
    print("Mean, median, and mode are not approximately equal.")
    print("The data may not follow a normal distribution.")
```

```python
# 3

import numpy as np

scores = np.random.normal(loc=70, scale=15, size=200)

within_one_sd = np.sum((scores >= 55) & (scores <= 85))

print("Number of players within one standard deviation:", within_one_sd)
print("Percentage:", (within_one_sd / 200) * 100)
```

```python
4. Real-world scenario — Zomato delivery times
Scenario: Zomato order delivery time

Normal distribution can be useful for modeling delivery times when most orders are delivered around an average time.
Most deliveries may fall close to the mean, while very fast or very slow deliveries occur less frequently.
If the delivery process is stable and affected by many small independent factors, the distribution may approximately follow a bell-shaped curve.
However, traffic, weather, restaurant delays, and peak hours can make the actual data skewed rather than perfectly normal.
```
