```python
# 1

import numpy as np

likes = np.array([
    [100, 150],
    [200, 250]
])

comments = np.array([
    [10, 20],
    [30, 40]
])

result = np.dot(likes, comments)

print("Likes Matrix:")
print(likes)

print("Comments Matrix:")
print(comments)

print("Resulting Matrix:")
print(result)
```

```python
# 3

def linear_regression_predict(x, m, c):
    return m * x + c


distances = [2, 5, 8]

m = 4
c = 10

for distance in distances:
    delivery_time = linear_regression_predict(distance, m, c)
    print("Distance:", distance, "km")
    print("Predicted delivery time:", delivery_time, "minutes")
```

4. Slope and intercept in Flipkart price prediction
Slope
The slope represents how much the predicted product price changes when the product rating increases by one unit. For example, if the slope is ₹500, the predicted price increases by ₹500 for every 1-point increase in rating.
Intercept
The intercept represents the predicted product price when the product rating is zero. It is the starting value of the regression equation before the effect of the rating is added.

5. Matrix multiplication in Spotify's recommendation system
Matrix multiplication helps a machine learning model combine information about users and songs. For example, one matrix can represent user preferences while another represents song features such as genre, artist, or popularity. Multiplying these matrices helps calculate how well a particular song matches a user's preferences. The system can then use these predicted scores to recommend songs that the user is likely to enjoy.
