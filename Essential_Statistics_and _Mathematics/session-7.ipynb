```python
import matplotlib.pyplot as plt

# Instagram usage hours per day
hours = [1, 2, 2.5, 3, 3.5, 4, 5]

# Number of posts per day
posts = [1, 2, 2, 3, 4, 5, 6]

# Create scatter plot
plt.scatter(hours, posts)

plt.xlabel("Hours spent on Instagram")
plt.ylabel("Number of posts")
plt.title("Instagram Usage vs Number of Posts")

plt.show()
```

```python
import numpy as np

# Monthly Zomato orders
zomato_orders = [5, 8, 10, 12, 15, 18]

# Monthly Swiggy orders
swiggy_orders = [3, 6, 8, 10, 13, 16]

# Calculate correlation coefficient
correlation = np.corrcoef(zomato_orders, swiggy_orders)[0, 1]

print("Correlation coefficient:", correlation)

if correlation > 0:
    print("The correlation is positive.")
elif correlation < 0:
    print("The correlation is negative.")
else:
    print("The correlation is zero.")
```

```python
import matplotlib.pyplot as plt

# Time spent on Spotify (hours)
spotify_hours = [1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5, 6]

# Test marks
marks = [95, 91, 88, 84, 81, 76, 72, 68, 63, 55]

# Create scatter plot
plt.scatter(spotify_hours, marks)

plt.xlabel("Time spent on Spotify (hours)")
plt.ylabel("Test Marks")
plt.title("Spotify Time vs Test Marks")

plt.show()
```
