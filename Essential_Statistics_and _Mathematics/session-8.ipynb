```python
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes


# Coin has 1 favorable outcome (Heads)
probability = calculate_probability(1, 2)

print("Probability of getting Heads:", probability)
```

```python
import random

count_six = 0
total_rolls = 100

# Roll the dice 100 times
for i in range(total_rolls):
    dice = random.randint(1, 6)

    if dice == 6:
        count_six += 1

# Calculate experimental probability
probability = count_six / total_rolls

print("Number of times 6 appeared:", count_six)
print("Experimental probability:", probability)
```

```python
genres = [
    'pop', 'rock', 'pop', 'hiphop', 'jazz',
    'pop', 'rock', 'pop', 'hiphop', 'pop',
    'rock', 'jazz', 'pop', 'rock', 'pop',
    'hiphop', 'jazz', 'pop', 'rock', 'pop'
]

pop_count = genres.count('pop')
total_songs = len(genres)

probability = pop_count / total_songs

print("Number of Pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of picking a Pop song:", probability)
```

```python
def probability_both(food_users, dessert_users):
    food_users = set(food_users)
    dessert_users = set(dessert_users)

    # Users who ordered both food and dessert
    both_users = food_users.intersection(dessert_users)

    # All unique users
    total_unique_users = food_users.union(dessert_users)

    probability = len(both_users) / len(total_unique_users)

    return probability


# Example data
food_users = [1, 2, 3, 4, 5, 6]
dessert_users = [3, 4, 5, 7, 8]

result = probability_both(food_users, dessert_users)

print("Probability of ordering both:", result)
```

```python
def conditional_probability(phone_users, headphone_users):
    phone_users = set(phone_users)
    headphone_users = set(headphone_users)

    # Users who bought both phone and headphones
    both_users = phone_users.intersection(headphone_users)

    # Probability of buying headphones given phone purchase
    probability = len(both_users) / len(phone_users)

    return probability


# Example users
phone_users = [1, 2, 3, 4, 5, 6, 7, 8]
headphone_users = [2, 3, 5, 9, 10]

result = conditional_probability(phone_users, headphone_users)

print("P(Headphones | Phone):", result)
```
