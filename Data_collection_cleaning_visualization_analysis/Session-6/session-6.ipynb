```python
# 2

import pandas as pd

df = pd.read_csv("ipl_player_stats.csv")

print("Shape before dropping missing values:")
print(df.shape)

clean_df = df.dropna(axis=0, how="any")

print("\nShape after dropping missing values:")
print(clean_df.shape)

print("\nCleaned DataFrame:")
print(clean_df)
```

```python
# 3

import pandas as pd

df = pd.read_csv("ipl_player_stats.csv")

df["runs"] = df["runs"].fillna(df["runs"].mean())
print("\nUpdated Runs Column:")
print(df["runs"])
```

```python
import pandas as pd

zomato_df = pd.DataFrame({
    "restaurant": [
        "Spice Hub",
        "Food Corner",
        "Pizza Palace",
        "Biryani House",
        "Taste Cafe",
        "Urban Kitchen",
        "Royal Restaurant"
    ],
    "rating": [
        4.5,
        None,
        None,
        4.2,
        None,
        4.7,
        None
    ]
})

print("Before Filling:")
print(zomato_df)

zomato_df["rating"] = zomato_df["rating"].ffill()

zomato_df["rating"] = zomato_df["rating"].bfill()

print("\nAfter Forward Fill + Backward Fill:")
print(zomato_df)
```

MAR (Missing At Random) is the most likely mechanism if missing runs are related to another observed variable, such as the number of matches played. For example, players with fewer matches may have incomplete statistics, so the missingness depends on information available in another column rather than being completely random.
