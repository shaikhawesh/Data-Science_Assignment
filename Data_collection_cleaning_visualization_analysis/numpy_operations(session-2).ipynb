```python
import numpy as np

# 1. Zomato ratings of 4 users for 5 food items
ratings = np.array([
    [4.5, 3.5, 5.0, 4.0, 3.0],
    [3.0, 4.5, 4.0, 5.0, 3.5],
    [5.0, 4.0, 3.5, 4.5, 5.0],
    [2.5, 3.0, 4.5, 3.5, 4.0]
])

second_third_users = ratings[1:3]
print("Ratings given by 2nd and 3rd users:")
print(second_third_users)
```

```python
# 2. Boolean indexing - select days with more than 8000 steps
steps = np.array([7500, 8200, 9000, 6500, 10500, 7800, 8500, 7200, 9200, 8100])

high_steps = steps[steps > 8000]

print("\nSteps greater than 8000:")
print(high_steps)
```

```python
# 3. Fancy indexing - select scores from matches 2, 5, and 7
ipl_scores = np.array([185, 210, 175, 198, 220, 165, 205, 190])


selected_scores = ipl_scores[[1, 4, 6]]

print("\nIPL scores from matches 2, 5, and 7:")
print(selected_scores)
```

```python
# 4. Broadcasting - apply a 10% discount
prices = np.array([500, 1000, 1500, 2000, 2500])

discounted_prices = prices * 0.90

print("\nOriginal prices:")
print(prices)

print("Prices after 10% discount:")
print(discounted_prices)
```

```python
# 5. Boolean masking - replace negative ratings with zero
song_ratings = np.array([-2, 5, 0, -1, 4, 3, -5, 2])

song_ratings[song_ratings < 0] = 0

print("\nSong ratings after replacing negative values with zero:")
print(song_ratings)
```
