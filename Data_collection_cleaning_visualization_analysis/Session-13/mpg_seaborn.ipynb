# Session 13 — MPG Dataset Visualization with Seaborn

This notebook explores Seaborn's built-in `mpg` dataset and creates six visualizations.

**Setup:** Install/update the required packages and ensure internet access is available the first time Seaborn downloads the dataset.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

OUTPUT_DIR = Path.cwd()
```

## Load and clean the MPG dataset

```python
mpg = sns.load_dataset("mpg")

# Remove rows with missing values
mpg = mpg.dropna().copy()

print("Cleaned dataset shape:", mpg.shape)
print(mpg.head())
```

## 1. Pairplot: MPG, horsepower, weight and acceleration

```python
g = sns.pairplot(
    data=mpg,
    vars=["mpg", "horsepower", "weight", "acceleration"],
    hue="origin",
    corner=True
)
g.fig.suptitle("MPG Relationships by Vehicle Origin", y=1.02)
g.savefig(OUTPUT_DIR / "mpg_pairplot.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 2. Correlation heatmap with upper triangle hidden

```python
corr = mpg.corr(numeric_only=True)

# True values in this mask hide the upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap of MPG Dataset")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "mpg_corr.png", dpi=150, bbox_inches="tight")
plt.show()

# Find the variable (other than mpg) with the strongest negative correlation with mpg
mpg_correlations = corr["mpg"].drop("mpg").dropna()
most_negative_column = mpg_correlations.idxmin()
print(
    "Most negatively correlated with mpg:",
    most_negative_column,
    f"(correlation = {mpg_correlations[most_negative_column]:.2f})"
)
```

## 3A. Figure-level line chart: MPG by model year and origin

```python
g = sns.relplot(
    data=mpg,
    x="model_year",
    y="mpg",
    hue="origin",
    kind="line"
)
g.set_axis_labels("Model Year", "Miles per Gallon (MPG)")
g.fig.suptitle("MPG by Model Year and Origin", y=1.03)
g.savefig(OUTPUT_DIR / "mpg_relplot.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3B. Figure-level bar chart: mean MPG by cylinders

```python
g = sns.catplot(
    data=mpg,
    x="cylinders",
    y="mpg",
    kind="bar",
    errorbar=("ci", 95)
)
g.set_axis_labels("Number of Cylinders", "Mean MPG")
g.fig.suptitle("Mean MPG by Number of Cylinders", y=1.03)
g.savefig(OUTPUT_DIR / "mpg_catplot.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3C. Figure-level distribution chart: MPG by origin

```python
g = sns.displot(
    data=mpg,
    x="mpg",
    hue="origin",
    kde=True
)
g.set_axis_labels("Miles per Gallon (MPG)", "Count")
g.fig.suptitle("MPG Distribution by Vehicle Origin", y=1.03)
g.savefig(OUTPUT_DIR / "mpg_displot.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 4. Joint plot: vehicle weight against MPG

```python
g = sns.jointplot(
    data=mpg,
    x="weight",
    y="mpg",
    kind="reg"
)
g.set_axis_labels("Vehicle Weight (lbs)", "Miles per Gallon (MPG)")
g.fig.suptitle("Vehicle Weight vs MPG", y=1.02)
g.savefig(OUTPUT_DIR / "mpg_joint.png", dpi=150, bbox_inches="tight")
plt.show()
```
