```python
# 1

import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:system@localhost:3306/tasks"
)

restaurants_df = pd.read_sql("restaurants", engine)

print("First 5 rows of restaurants table:")
print(restaurants_df.head())
```

```python
# 2

query = """
SELECT song_name, artist
FROM musicplaylist
WHERE duration > 200
"""

song_df = pd.read_sql_query(query, engine)

print("Music with duration greater than 200:")
print(song_df)
```

```python
# 3

import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"

users_df = pd.read_json(url)

print("Usernames:")
print(users_df["username"])
```

```python
# 4

import pandas as pd

orders_df = pd.read_csv("orders.csv")
users_df = pd.read_csv("users.csv")

combined_df = pd.merge(
    orders_df,
    users_df,
    on="user_id",
    how="inner"
)

result = combined_df[["order_id", "username", "amount"]]

print("Combined Orders and Users:")
print(result)
```

```python
# 5

import pandas as pd

today_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "item": ["Pizza", "Burger", "Biryani"],
    "price": [300, 200, 250]
})

yesterday_orders = pd.DataFrame({
    "order_id": [98, 99, 100],
    "item": ["Pasta", "Sandwich", "Dosa"],
    "price": [220, 150, 120]
})

combined_orders = pd.concat(
    [today_orders, yesterday_orders],
    ignore_index=True
)

combined_orders = combined_orders.reset_index(drop=True)

print("Combined Orders:")
print(combined_orders)
```
