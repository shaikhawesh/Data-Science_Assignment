```python
# 1

import pandas as pd

df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "restaurant_name": [
        "Dominos",
        "Pizza Hut",
        "McDonalds",
        "Dominos",
        "KFC",
        "Pizza Hut",
        "Dominos"
    ],
    "order_date": [
        "2024-06-01",
        "2024-06-02",
        "2024-06-03",
        "2024-06-01",
        "2024-06-05",
        "2024-06-02",
        "2024-06-10"
    ]
})

print("Original Zomato Orders:")
print(df)

duplicates = df[df.duplicated(
    subset=["restaurant_name", "order_date"],
    keep=False
)]

print("\nDuplicate Orders:")
print(duplicates)
```

```python
# 4

import pandas as pd

df = pd.DataFrame({
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "rahul123",
        "insta_queen",
        "priya456",
        "insta-queen"
    ]
})

print("Original Usernames:")
print(df)

df["username"] = df["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen"
})

print("\nStandardized Usernames:")
print(df)
```

```python
# 5

import pandas as pd

df = pd.DataFrame({
    "user": [
        "Rahul",
        "Priya",
        "Amit",
        "Neha",
        "Raj",
        "Karan",
        "Anjali",
        "Vishal"
    ],
    "payment_status": [
        "Yes",
        "yes",
        "Y",
        "No",
        "no",
        "N",
        " Yes ",
        " No "
    ]
})

print("Original Payment Status:")
print(df)

df["payment_status"] = df["payment_status"].str.strip().str.lower()

df["payment_status"] = df["payment_status"].map({
    "yes": 1,
    "y": 1,
    "no": 0,
    "n": 0
})

print("\nUpdated Payment Status:")
print(df)

print("\nData Type:")
print(df["payment_status"].dtype)
```
