```python
# 1

import pandas as pd

timestamps = [
    '2024-06-01 14:30',
    '2024-06-02 09:15',
    '2024-06-03 20:45'
]

datetime_values = pd.to_datetime(timestamps)

print("Converted Datetime Values:")
print(datetime_values)
```

```python
# 2

import pandas as pd

df = pd.read_csv("flipkart_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["weekday"] = df["order_date"].dt.day_name()

print(df)
```

```python
# 3

import pandas as pd

df = pd.read_csv("flipkart_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df = df.set_index("order_date")

weekly_orders = df.resample("W").size()

print("Weekly Order Count:")
print(weekly_orders)
```

```python
# 4 

import pandas as pd

df = pd.DataFrame({
    "post_id": [1, 2, 3, 4, 5],
    "posted_at": [
        "2024-06-01 08:30:00+00:00",
        "2024-06-02 12:15:00+00:00",
        "2024-06-03 15:45:00+00:00",
        "2024-06-04 10:00:00+00:00",
        "2024-06-05 18:30:00+00:00"
    ]
})

df["posted_at"] = pd.to_datetime(df["posted_at"], utc=True)

df["posted_at_india"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")

print(df["posted_at_india"].head())
```

```python
# 5

import pandas as pd

df = pd.read_csv("flipkart_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

print(df[["order_date", "is_weekend"]])
```
