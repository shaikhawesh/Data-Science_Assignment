```python
"""
1.Install the sqlite3 module in Python and write a script to create a
 new database called foodie.db with a table Restaurants 
 (id, name, cuisine, rating)."""

import sqlite3

# Connect to database
connection = sqlite3.connect("foodie.db")

# Create cursor
cursor = connection.cursor()

# Create Restaurants table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Restaurants (
    id INTEGER PRIMARY KEY,
    name TEXT,
    cuisine TEXT,
    rating REAL
)
""")

# Save changes
connection.commit()

# Close connection
connection.close()

print("Database and table created successfully.")
```

```python
"""
2.Using sqlite3 in Python, insert three sample restaurants into the Restaurants
 table in foodie.db and write a query to fetch all restaurants with a
   rating above 4.0, then print their names.
"""

import sqlite3

connection = sqlite3.connect("foodie.db")
cursor = connection.cursor()

# Insert restaurants
cursor.executemany("""
INSERT INTO Restaurants (id, name, cuisine, rating)
VALUES (?, ?, ?, ?)
""", [
    (1, "Swagat Restaurant", "Indian", 4.5),
    (2, "Pizza Palace", "Italian", 4.2),
    (3, "Dragon House", "Chinese", 3.9)
])

connection.commit()

# Fetch restaurants with rating above 4.0
cursor.execute("""
SELECT name
FROM Restaurants
WHERE rating > 4.0
""")

restaurants = cursor.fetchall()

# Print restaurant names
for restaurant in restaurants:
    print(restaurant[0])

connection.close()
```

```python
"""
4.Add a new column 'delivery_charge' to your DataFrame, setting it to 50 
for all restaurants, and then calculate a new column 'final_rating' as rating + (0.1 if cuisine is 'Italian').
<br><br><em><strong>Hint:</strong> 
Use DataFrame.apply() or a lambda function for the conditional logic.</em>
"""

import sqlite3
import pandas as pd

connection = sqlite3.connect("foodie.db")

df = pd.read_sql_query(
    "SELECT * FROM Restaurants",
    connection
)

# Add delivery charge
df["delivery_charge"] = 50

# Calculate final rating
df["final_rating"] = df.apply(
    lambda row: row["rating"] + 0.1
    if row["cuisine"] == "Italian"
    else row["rating"],
    axis=1
)

print(df)

connection.close()
```

```python
"""
5.
Automate a daily summary: Write a Python script that 
connects to foodie.db, fetches all restaurants with rating above 4.5,
 loads them into a DataFrame,
 and saves the result as a CSV file named top_rated_restaurants.csv.
"""

import sqlite3
import pandas as pd

# Connect to database
connection = sqlite3.connect("foodie.db")

# Fetch restaurants with rating above 4.5
query = """
SELECT *
FROM Restaurants
WHERE rating > 4.5
"""

# Load results into DataFrame
df = pd.read_sql_query(query, connection)

# Save DataFrame as CSV
df.to_csv(
    "top_rated_restaurants.csv",
    index=False
)

# Close database connection
connection.close()

print("Daily summary saved successfully.")
```
